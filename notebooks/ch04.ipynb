```python
# p86
weight = 0.5
input = 0.5
goal_prediction = 0.8

step_amount = 0.001
for interation in range(1101):
    prediction = input * weight
    error = (prediction - goal_prediction) ** 2

    print(f"Error: {str(error)} / Prediction: {str(prediction)}")

    up_prediction = input * (weight + step_amount)
    up_error = (goal_prediction - up_prediction) ** 2

    down_prediction = input * (weight - step_amount)
    down_error = (goal_prediction - down_prediction) ** 2

    if (down_error < up_error):
        weight = weight - step_amount

    if (down_error > up_error):
        weight = weight + step_amount
```

```python
import numpy as np
import matplotlib.pyplot as plt

weight, goal_pred, input_val = 0.0, 0.8, 1.1

# 오차 함수
weights = np.linspace(-0.28, 1.45, 500)
errors = (input_val * weights - goal_pred) ** 2

# 현재 위치와 기울기
pred = input_val * weight
error = (pred - goal_pred) ** 2
slope = 2 * (pred - goal_pred) * input_val

# 실제 접선: 검은 점을 기준으로 좌우 모두 표시
tangent_x = np.linspace(-0.27, weight + 0.30, 100)
tangent_y = error + slope * (tangent_x - weight)

# 세로가 더 긴 차트
fig, ax = plt.subplots(figsize=(4.2, 5.5), dpi=110)

# 오차 곡선
ax.plot(weights, errors, color="#28796d", linewidth=2.2)

# 검은 점선 접선
ax.plot(
    tangent_x, tangent_y,
    color="black",
    linestyle=(0, (1, 3)),
    linewidth=2,
    zorder=3
)

# 현재 위치
ax.scatter(weight, error, color="black", s=240, zorder=5)

# 축 범위
ax.set_xlim(-0.30, 1.45)
ax.set_ylim(0, 1.25)

# 책처럼 축을 원점에 배치
ax.spines["left"].set_position(("data", 0))
ax.spines["bottom"].set_position(("data", 0))
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.set_xticks([])
ax.set_yticks([])

ax.set_xlabel("weight = 0.0", fontsize=12)
ax.set_ylabel("error = 0.64", fontsize=12, labelpad=12)

ax.grid(False)

plt.tight_layout()
plt.show()
```
